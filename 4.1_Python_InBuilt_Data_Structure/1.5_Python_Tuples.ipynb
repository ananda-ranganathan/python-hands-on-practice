```python
# Tuples in Python
# A tuple is a collection which is ordered and unchangeable.
# Tuples are written with round brackets.
# Tuple items are ordered, unchangeable, and allow duplicate values.
# Tuples are immutable
# They are similar to lists, but thier immutability makes them different.
# Tuples are used to store multiple items in a single variable.
```

```python

# Creating a Tuple
empty_tuple = ()
print(empty_tuple)
print(type(empty_tuple))
```

```python
tuple_list = list()
print(type(tuple_list))
tpl = tuple()
print(type(tpl))
```

```python
# Creating Elements inside the Tuple
tpl_numbers = tuple([1,1,2,3,4,5])
print(tpl_numbers)

list((1,1,2,3,4,5,6,7))

# Mixed Datatyples inside the Tuple
tpl_mixed = (1,"Hello",3.4,False)
print(tpl_mixed)
```

```python
# Accessing elements inside the tuple
tpl_acc_elements = (1,"Anand",30,4.5,"Erode")
print(tpl_acc_elements)

print(tpl_acc_elements[1])
print(tpl_acc_elements[-2])

# Accessing the range of indexes (slicing Operation)
print(tpl_acc_elements[1:4])
print(tpl_acc_elements[::])
print(tpl_acc_elements[::-1])

```

```python
# Tuples Oprations 
tpl_op = (6, "Ananda")
tpl_mix_op = ("Ranganathan","@Data Scientist")

# Concatination
Concat_tpl = tpl_op + tpl_mix_op
print(Concat_tpl)
```

```python
# Calculation with Tuples 
append_calc = Concat_tpl * 3 # THIS WILL PRINT/APPEND THE TUPLE LIST THREE TIMES
print(append_calc)
```

```python

# This below example is List which is changeable 
List  = [1,2,3,4,5]
print(List)
# Updating the he value of List Index 2
List[2] = "8"
print(List)

# immutable Nature of Tuples 
# Tuples are immutable means we cannot change the values of the tuple once it is created.

unchanged_tp = (1,2,3,4,5,6,6)
print(unchanged_tp)

# updating the elemnet Value will get error
unchanged_tp[2] = "Hello"
print(unchanged_tp)
```

```python

# Tuple Methods
tpl_mthd = (1,2,3,4,4,4,45)
print(tpl_mthd)
# count() - Returns the number of times a specified value appears in the tuple.
print(tpl_mthd.count(45)) # (45) - is the value which is to be searched in the tuple not the index

# index() - Returns the index of the first element with the specified value.
#print(tpl_mthd.index(5)) # (5) - is the value which is to be searched in the tuple not the index - Throes an error
print(tpl_mthd.index(45))


# tuple() - Converts the specified iterable to a tuple.
# list() - Converts the tuple to a list.

```

```python
# Packing and Unpacking
# Packing - Assigning multiple values to multiple variables
# Unpacking - Assigning one value to multiple variables

# Packing 
packed_tpl = 1,"Anand",3.14,"Erode"
print(packed_tpl)

# Lets
x = 1
y = "Anand"
z = 3.14
city = "Erode"

# Unpacking

unpacked_tpl = x,y,z,city
print(unpacked_tpl)
```

```python
# Unpacking with *(star)
numbers=(1,3,4,5,6,7,8,9,0)
first,*middle,last = numbers
print(first)
print(middle)
print(last)
```

```python
# nested List
nested_list = [[1,2,3,4],[6,7,8,9],[5,"Anand",3.14,"Erode"]]
# Slicing Operation allowed inside the Nested List
print(nested_list[0][1:3])
print(nested_list[1][1:3])
nested_list[2][1:3]  # [2] - Index of the Nested List ; [1:3] - Value iNDEXES inside the Nested List
```

```python
# Nested Tuple
nested_tpl = ((1,2,3,4),(6,7,8,9),(5,"Anand",3.14,"Erode"))
print(nested_tpl)

# accessing elements inside the nested tuple
print(nested_tpl[0])
print(nested_tpl[1][3])  # [1] - Index of the Nested Tuple ; [3] - Value INDEX inside the Nested Tuple

# Slicing Operation allowed inside the Nested Tuple
print(nested_tpl[0][1:3])
print(nested_tpl[1][1:3])

```

```python
# Iterating Over Nested Tuples=
Iterate_tuple = ((1,2,3,4),(6,7,8,9),(5,"Anand",3.14,"Erode")) 
for sub_tuple in Iterate_tuple: 
    for item in sub_tuple: 
        print(item,end=" ") 
    print()

# Explain the abobe code how looping works
# for sub_tuple in Iterate_tuple: 
#     for item in sub_tuple: 
#         print(item,end=" ") 
#     print()

# 1. Iterate_tuple = ((1,2,3,4),(6,7,8,9),(5,"Anand",3.14,"Erode"))
    # A variable named Iterate_tuple is created.
    # It stores a tuple of tuples:
        # First sub-tuple: (1,2,3,4)
        # Second sub-tuple: (6,7,8,9)
        # Third sub-tuple: (5,"Anand",3.14,"Erode")

# 2.  for sub_tuple in Iterate_tuple:
        # Outer loop: iterates through each sub-tuple inside Iterate_tuple.
        # On each iteration, sub_tuple takes one of the inner tuples.

# 3. for item in sub_tuple:
        # Inner loop: iterates through each element inside the current sub_tuple.
        # On each iteration, item takes one element of the tuple.

# 4.  print(item,end=" ")
        # Prints the current item.
        # end=" " ensures items are printed on the same line separated by spaces (instead of new lines).

# 5. print()
        # After finishing the inner loop for one sub-tuple, this prints a newline.
        # This Ensures each sub-tuple’s items appear on a separate line.

# 📊 Trace Table
# Iteration (Outer Loop)	sub_tuple Value	      Inner Loop Iterations (item)                  	Printed Output (line)
#          1	                 (1,2,3,4)	        1 → 1, 2 → 2, 3 → 3, 4 → 4	                         1 2 3 4
#          2		             (6,7,8,9)          1 → 6, 2 → 7, 3 → 8, 4 → 9	                         6 7 8 9
#          3	         (5,"Anand",3.14,"Erode")   1 → 5, 2 → "Anand", 3 → 3.14, 4 → "Erode"	      5 Anand 3.14 Erode

# ✅ Final Output
# Code
# 1 2 3 4 
# 6 7 8 9 
# 5 Anand 3.14 Erode 



```
