```python
# Sets in python -
# Definition: Sets are unordered collections of unique elements. 
# They are similar to lists, but they do not allow duplicate elements.
# They are mutable, meaning their elements can be added or removed after creation.
# They are also iterable, allowing you to iterate over their elements using a for loop.

# Creating a set
my_set = {1, 2, 3}
print(my_set)

# Accessing elements in a set
print(my_set[0])  # This will raise an error, sets are not indexed

# Adding elements to a set
my_set.add(4)
print(my_set)

# Removing elements from a set
my_set.remove(2)
print(my_set)

# Iterating over a set
for element in my_set:
    print(element)

# Set operations
set1 = {1, 2, 3}
set2 = {3, 4, 5}
union_set = set1.union(set2)
intersection_set = set1.intersection(set2)
difference_set = set1.difference(set2)
symmetric_difference_set = set1.symmetric_difference(set2)
print(union_set)
print(intersection_set)
print(difference_set)
print(symmetric_difference_set)

# Membership check
print(3 in set1)
print(6 in set1)

# Length of a set
print(len(set1))

# Clearing a set
set1.clear()
print(set1)

# Frozen set
frozen_set = frozenset([1, 2, 3])
print(frozen_set)

# Set Comprehension
numbers = [1, 2, 3, 4, 5]
squares = {num**2 for num in numbers}
print(squares)

# Nested Set
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set3 = {5, 6, 7}
nested_set = {set1, set2, set3}
print(nested_set)

# Set Methods
set1 = {1, 2, 3}
set2 = {3, 4, 5}
set1.update(set2)
print(set1)

set1.intersection_update(set2)
print(set1)

set1.difference_update(set2)
print(set1)

set1.symmetric_difference_update(set2)
print(set1)

set1.discard(3)
print(set1)

set1.remove(2)
print(set1)

set1.pop()
print(set1)

set1.clear()
print(set1)
```

```python

# Mathematical Operations
# Definition : Mathematical operations can be performed on sets

my_set1 = {1,2,3,4,5,6}
my_set2 = {3,4,5,6,7,8}

# Union - The union of two sets is the set of all elements that are in either set.
print(my_set1.union(my_set2)) 
# Intersection - The intersection of two sets is the set of all elements that are in both sets.
print(my_set1.intersection(my_set2))
# Intersection_Update - The intersection update of two sets is the set of all elements 
# that are in both sets.
print(my_set1.intersection_update(my_set2))




# my_set1 = {1,2,3,4,5,6}
# my_set2 = {3,4,5,6,7,8}
print(my_set1)
print(my_set2)
```

```python

my_set1 = {1,2,3,4,5,6}
my_set2 = {3,4,5,6,7,8}

# Difference - The difference of two sets is the set of all elements that are in the first set but not in the second set.
print(my_set1.difference(my_set2))
# Difference_Update - The difference update of two sets is the set of all elements that are in the first set but not in the second set.
print(my_set1.difference_update(my_set2))
# Symmetric Difference - The symmetric difference of two sets is the set of all elements that are in either set 
# but not in both sets.           
print(my_set1.symmetric_difference(my_set2)) 




print(my_set1)
print(my_set2)
```

```python

my_set1 = {1,2,3,4,5,6}
my_set2 = {3,4,5,6,7,8}

# Subset - The subset of two sets is the set of all elements that are in the first set and also in the second set.
print(my_set1.issubset(my_set2))
# Superset - The superset of two sets is the set of all elements that are in the first set and also in the second set.
print(my_set1.issuperset(my_set2))  

```
