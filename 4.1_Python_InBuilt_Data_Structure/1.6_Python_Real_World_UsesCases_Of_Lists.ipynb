```python
# Input data: contains duplicates
data = [1, 2, 2, 3]

# -------------------------------
# LIST
# -------------------------------
# Lists are ordered, mutable, and allow duplicates
my_list = data
print("List:", my_list)  # preserves duplicates and order

# Lists can be modified (mutable)
my_list.append(4)        # add new element at the end
print("List after append:", my_list)

# -------------------------------
# TUPLE
# -------------------------------
# Tuples are ordered, immutable, and allow duplicates
my_tuple = tuple(data)
print("Tuple:", my_tuple)  # preserves duplicates and order

# Tuples cannot be modified, but support read-only methods
print("Tuple count of 2:", my_tuple.count(2))  # count occurrences
print("Tuple index of 3:", my_tuple.index(3))  # find position

# -------------------------------
# SET
# -------------------------------
# Sets are unordered, mutable, and store only unique elements
my_set = set(data)
print("Set:", my_set)  # duplicates removed automatically

# Sets can be modified (mutable)
my_set.add(4)          # add new element
print("Set after add:", my_set)

# -------------------------------
# DICTIONARY
# -------------------------------
# Dictionaries are key-value mappings, keys must be unique
my_dict = {"a": 1, "b": 2, "c": 2, "d": 3}
print("Dictionary:", my_dict)  # values can repeat, keys unique

# Access values by key
print("Access key 'b':", my_dict["b"])

# Dictionaries can be modified (mutable)
my_dict["e"] = 4       # add new key-value pair
print("Dictionary after adding key 'e':", my_dict)

# Expected Output:
# List: [1, 2, 2, 3]
# List after append: [1, 2, 2, 3, 4]

# Tuple: (1, 2, 2, 3)
# Tuple count of 2: 2
# Tuple index of 3: 3

# Set: {1, 2, 3}
# Set after add: {1, 2, 3, 4}

# Dictionary: {'a': 1, 'b': 2, 'c': 2, 'd': 3}
# Access key 'b': 2
# Dictionary after adding key 'e': {'a': 1, 'b': 2, 'c': 2, 'd': 3, 'e': 4}

```

```python
# Interview Takeaway
# List → ordered, mutable, duplicates allowed.

# Tuple → ordered, immutable, duplicates allowed.

# Set → unordered, mutable, duplicates removed.

# Dictionary → unordered mapping, keys unique, values can repeat.
```
