## Basic List Comprehension - Very Important Slide
Basic Syntax [Expression for items in iterable ] <br>
Also we can add with Conditional Logic [expression(Left Side) For item in Iterable (Middle) if condition (right most)] <br>
Nested List Comprehension - [Expresion for item 1 in iterable 1 for item 2 in iterable 2]

```python
# List Comprehension
# Basic Syntax : [expression for item in iterable]
square = [num **2 for num in range(10)]
print(square)
# With Logics and Conditions : [expression for item in iterable if condition]
Even_Number = [ num for num in range(10) if num % 2 == 0] 
# Create a list of numbers for each num in range 0 to 9 only if num is even”
print(Even_Number)

# Nested Comprehensive List - 
lst1 = [1,2,3,4,5]
lst2 = ['a','b','c','d','e']

pair = [[i,j] for i in lst1 for j in lst2]
print(pair)


```
