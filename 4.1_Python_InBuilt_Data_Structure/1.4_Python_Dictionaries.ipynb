```python
# Python Dictionaries - A collection of key-value pairs.
# Dictionaries are used to store data values in key:value pairs.
# A dictionary is a collection which is ordered*, changeable and does not allow duplicates.
# Dictionaries are written with curly brackets, and they have keys and values.
# A key is like an index that points to a value.
# 
# In Dictionaries, the key is unique and immutable(String,number,tuple).(that cannot be changed once assigned).
# In Disctionaries, the value can be mutable(any type of data type).(that can be changed anytime after assignation).

```

```python
# Creating Empty Dictionary
empty_dict = {}
print(type(empty_dict))

# Creating Empty Dictionary with dict Class and initiallizing it as method()
empty_dict = dict()      - # Empty Dictionary , the yelloe font represents the class dict and the () font represents the method dict()
print(type(empty_dict)) 
```

```python
# Dict_Key Value Pair
student = {"Name-":"Anand","Age-": 28,"Gender-":"Male"} 
# Where, "Name-" is key and "Anand" is value; "Age-" is key and "28" is value; "Gender-" is key and "Male" is value.
# The complete set pair is called Dictionary
print(type(student),student)                            
```

```python
# Error /mistakes in dictionaries
# Make Sure to use always SINGLE KEY ( Do Not create the key with the same string or text or numbers, 
# as is replace the duplicate key's value to the prevous key's value.)
Student = {"name" : "anand", "age" : 28, "gender" : "male", "name": "Ananda Ranganathan"}
# In above dict, the name key is assigned twice, name : anand and name : Ananda Ranganathan, 
# so dictionary will replace latest value (Ananda Ranganathan) to the previous value (anand) 
# Run and Check in the output
print(type(Student),Student)
```

```python
# Accessing dictionaries Elements 
STUDENT = {"Name" : "Anand", 'Age' : 28, "Gender" : "Male", "Profession" : "Learning Data Science"}
print(type(STUDENT))
print(STUDENT)

# Accessing the dictionary elements
# 1. Using keys()
print(STUDENT['Age'])
print(STUDENT["Profession"])

# 2. Using get() method
print(STUDENT.get("Name"))
print(STUDENT.get("Last_Name")) # If the key is not found in the dictionary, the get method will return none
print(STUDENT.get("Last_Name","Not Avaiable in the dictionary")) # If the key is not found in the dictionary, 
# the get method will return a message given by the user, "Not Avaiable in the dictionary"

# 3. Using values()
```

```python
# Modifying the dictionary elements
# Dictionaries are mutable,so we can modify - Add, Update, Remove or delete the elements
STUDENT_s = {"Name" : "Anand", 'Age' : 28, "Gender" : "Male", "Profession" : "Learning Data Science"}
print(type(STUDENT_s))
print(STUDENT_s)

# adding New Key and Value
STUDENT_s["Last_Name"] = "Ranganathan"
print(STUDENT_s)

# updating the value of the existing key
STUDENT_s["Name"] = "Ananda"
print(STUDENT_s)

# removing the key and value
del STUDENT_s["Age"]
print(STUDENT_s)
```

```python
  # 2. copy() - Returns a copy of the dictionary, shallow copy provides a different memory location so the original dictionary variable will not be affected
Student_Original = {"Name" : "Anand", 'Age' : 31, "Gender" : "Male", "Profession" : "Learning Data Science"}
print(type(Student_Original))
print(Student_Original)

# Copy the dictionary
Student_Shallow_Copy = Student_Original.copy()
print(type(Student_Shallow_Copy))
print(Student_Shallow_Copy)

# Updating the shallow copy name - value to ranganathan
Student_Shallow_Copy["Name"] = "Ranganathan"
print(Student_Original)
print(Student_Shallow_Copy)
```

```python
## 14 - Nested Dictionaries
# A nested dictionary is a dictionary that contains other dictionaries as its values.
# A nested dictionary is a dictionary that contains other dictionaries as its values.
Student_Nested = {
    'Student_Nested_Dict1' : {"Name" : "Anand", 'Age' : 31, "Gender" : "Male", "Profession" : "Learning Data Science"},
    'Student_Nested_Dict2' : {"Address" : "20/1, Yamunai Street", 'City' : "Erode", "State" : "Tamil Nadu", "Country" : "India"}
}
print(type(Student_Nested))
print(Student_Nested)

# Access Nested Dictionaries Elements 
print(Student_Nested['Student_Nested_Dict1'])
print(Student_Nested['Student_Nested_Dict2'])

# Correct ways to access nested values:
print(Student_Nested['Student_Nested_Dict1']['Name'])                # access Name inside first nested dict
print(Student_Nested['Student_Nested_Dict2']['City'])                # access City inside second nested dict
print(Student_Nested['Student_Nested_Dict2']['Country'])               # access Country inside second nested dict

# Access City and Country in same print statement
print(f"{Student_Nested['Student_Nested_Dict2']['City']}, {Student_Nested['Student_Nested_Dict2']['Country']}")

```

```python
# Iterate over Nested Dictionaries 
Student_ID = {"Name" : "Anand", 'Age' : 31, "Gender" : "Male", "Profession" : "Learning Data Science"}
Student_Info = {"Address" : "20/1, Yamunai Street", 'City' : "Erode", "State" : "Tamil Nadu", "Country" : "India"}

# Iterate Over Nested Dictionaries
for key, value in Student_ID.items():
    print(f"{key}: {value}")
for key, value in Student_Info.items():
    print(f"{key}: {value}")
```

```python
# Comprehensive Dictionary
Squrs = {x:x**2 for x in range(5)}
print(Squrs)

# Conditinal Dictionary Comprehension
evens = {x:x**2 for x in range(10) if x%2==0}
print(evens)
```
