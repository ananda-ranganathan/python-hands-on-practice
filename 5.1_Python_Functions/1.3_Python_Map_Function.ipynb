```python
# Map Function in Python

# Python map() function returns a map object(which is an iterator) of the results after applying the given function 
# to each item of a given iterable (list, tuple etc.).  
# The map() function is similar to the for loop, but is much more efficient and can also be used with anonymous functions.

# Syntax
# map(function, iterable(s))

# Parameters
# function -- It is a function to which map passes each element of given iterable.
# iterable -- It is a sequence of elements which is passed to function.

# Return Value -- It returns a map object.
```

```python
def squares(x):
    return(x*x)

print(squares(2))

# Squares using Map with List of numbers with above function name squares
num = [1,2,3,4,5,6,7,8]
# map(squares,num) --> creating map object at specific memory location 
# list(map(squares,num)) --> creating list of map object; 
list(map (squares ,num))
```

```python
# Squares using Lambda function with Map
num = [1,2,3,4,5,6,7,8]
# map(squares,num) --> creating map object at specific memory location 
# list(map(squares,num)) --> creating list of map object; 
sq=list(map(lambda num : num*num,num)) 
print(sq)
```

```python
# Multiple Iterable in map()
numbers1 = [1,2,3,4,5]
numbers2 = [6,7,8]

added_numbers= map(lambda x,y : x+y, numbers1, numbers2)
print(added_numbers)

added_numbers= list[map(lambda x,y : x+y, numbers1, numbers2)]
print(added_numbers)

added_numbers= list(map(lambda x,y : x+y, numbers1, numbers2))
print(added_numbers)
```

```python
# Map() to convert a list of strings to integers

# Use Maps to convert strings to integers

str_numbers = ['1','2','3','4','5','6']

int_numbers = list(map(int,str_numbers))
print(int_numbers)

# Use Maps to convert int to string
int_numbers = [3,4,5,6,7,8,9]

str_numbers = list(map(str,int_numbers))
print(str_numbers)
```

```python
# Inbuilt functions of MAPS

#str.upper()
words = ['apple','banana','pine-apple','cherry']
upper_words= list(map(str.upper,words))
print(upper_words)


# example of one dictionary
def get_name(person):
    return person['name'] #,person['age']

# List of dictionaries 
people =[
    {'name':"Anand",'age': 20},
    {'name':"Jack",'age': 30}
]
list(map(get_name,people))

```

```python
# 🧩 What is map() in Python?
# Definition: map() is a built‑in Python function that applies another function to every item of an iterable (like a list, tuple, or string) and returns a map object (an iterator).

# Think of it as: "Apply this function to each element in that collection."

# 📜 Syntax
# python
# map(function, iterable)
# function → the operation you want to perform (can be built‑in or user‑defined).

# iterable → the sequence of elements (list, tuple, set, etc.).

# ⚡ Real‑time Short Example (with word‑by‑word explanation)
# python
# numbers = [1, 2, 3, 4]
# squared = map(lambda x: x**2, numbers)
# print(list(squared))
# Word by word:
# numbers = [1, 2, 3, 4] → our iterable (list of integers).

# map(...) → tells Python: "I want to apply a function to each element."

# lambda x: x**2 → the function: take x and return x squared.

# numbers → the iterable to process.

# squared = map(...) → creates a map object (not a list yet).

# list(squared) → converts the map object into a list so we can see results.

# Output → [1, 4, 9, 16].

# 🗂️ How Map Object is Stored and Where?
# map() returns a map object, which is an iterator.

# It doesn’t store results immediately; it computes values on demand (lazy evaluation).

# To see it visually, convert it into a list, tuple, or set:

# python
# list(map(lambda x: x+10, [1,2,3]))
# # Output: [11, 12, 13]
# Think of it as a pipeline: the function is applied only when you iterate over the map object.

# 🎯 Why is map() Used?
# To avoid writing explicit loops.

# Makes code shorter, cleaner, and faster.

# Useful when applying the same transformation to all elements.

# ⏱️ When to Use map()?
# When you want to transform each element of a sequence.

# When performance matters (lazy evaluation avoids unnecessary memory usage).

# Example: processing large datasets, applying mathematical operations, cleaning strings.

# 🌟 Features of map()
# Functional style: encourages functional programming.

# Lazy evaluation: computes values only when needed.

# Works with any iterable: lists, tuples, sets, strings.

# Can use built‑in or custom functions.

# Returns an iterator: memory efficient.

# 🔧 Functions and Iterables in map()
# Function: can be built‑in (str.upper, abs, len) or user‑defined (lambda, def).

# Iterable: list, tuple, set, string, dictionary keys/values.

# Example:

# python
# words = ["hello", "world"]
# uppercased = map(str.upper, words)
# print(list(uppercased))  # ['HELLO', 'WORLD']
# 📚 Inbuilt Functions Commonly Used with map()
# Here are some popular ones:

# Function	Example	Output
# str.upper	list(map(str.upper, ["a","b"]))	['A','B']
# abs	list(map(abs, [-1,-2,3]))	[1,2,3]
# len	list(map(len, ["hi","python"]))	[2,6]
# int	list(map(int, ["1","2","3"]))	[1,2,3]
# float	list(map(float, ["1.1","2.2"]))	[1.1,2.2]
# round	list(map(round, [1.234, 2.678], [1,2]))	[1.2, 2.68]
# 🖼️ Visual Trace (How map() works internally)
# Imagine map(lambda x: x*2, [1,2,3]):

# Step	Element	Function Applied	Result
# 1	1	1*2	2
# 2	2	2*[2](guide://action?prefill=Tell%20me%20more%20about%3A%202)	4
# 3	3	3*2	6
# Final output → [2,4,6].

# ✅ So in interviews, you can confidently say:

# "map() applies a function to each element of an iterable, returns a lazy iterator, and is often used with built‑in functions or lambdas for clean transformations."
```
