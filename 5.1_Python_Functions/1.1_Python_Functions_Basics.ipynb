```python

# Reuse the above entire code inside the function for reusability
# Define or create the even or odd function 
def even_or_odd(num): # def is the keyword for defining the function ; even_or_odd is the name of the function ; num is the parameter
    # docstring : the function will check if the number is even or odd
    if num%2==0:
        print("the number s even")
    else:
        print("the number is odd")
    
```

```python
# Function with multiple parameters
# defining (Add two numbers) function name 
def add_2_num (num1,num2): # num1 and num2 are the two different parameters
    # docstring: the function will add two numbers
    return (num1+num2)

#caling the function
result = add_2_num(2,4) 
# result is the new variable created to store the return value of the function;
# add_2_num is the function and 2 and 4 are the arguments
print(result) # print the result
 
# OR

add_2_num(2,3) # add_2_num is the function and 2 and 3 are the arguments
```

```python
# Default Parameters within our fucntion
def greet(name = 'Guest'):
    print(f"hello {name}, Welcome to the function learning")
# calling the greet fucntions
greet() # greet is the function and passing 'nil'argument

# Note: 
# Parameters ('name') are the variables inside the function and Guest is the default arument value
# Default parameters are the parameters that are assigned a default value if no argument is provided when the function is called.
```

```python
# Default Parameters within our fucntion
def greet(name):
    print(f"hello {name}, Welcome to the function learning")
# calling the greet fucntions
greet('Anand') # greet is the function and passing "anand" is the argument

# Arguments ('Anand') are must important to provide when calling the function
# In the above example, the name parameter has a default value of "Anand" if no name is provided when the function is called.
# If a name is provided when the function is called, it will override the default value.
# This is useful when you want to provide a default value for a parameter that is not always required when the function is called.
```

```python
# variable Length Argumments : 
# Postional Argumments and Keyword Argumments : *args and **kwargs 

# Postional Arguments 
# def print_numbers(num1,num2,num3,num4,num5,num6,num7):
#     print(num1)
#     print(num2)
#     print(num3)
#     print(num4)
#     print(num5)
#     print(num6)
#     print(num7)
# print_numbers(1,2,3,4,5,6,7)

# instead of providing num1,num2,num3,num4,num5,num6,num7 we can use *args
def print_numbers(*args):
    # docstring: print_number is the function and *args are the postional arguments
    # *args stores all positional arguments as a tuple
    for nums in args: # for loop to print the numbers ; nums is the variable created to store the numbers
        print(nums)

# Calling the print_number fucntion
print_numbers(1,2,3,4,5,6,"Anand",7)
    
```

```python
# Keywords arguments 
def print_kwargs(**kwargs):
    # docstring: print_kwargs is the function and **kwargs are the keyword arguments
    # **kwargs stores all keyword arguments as a dictionary
    for key, value in kwargs.items():
        print(f"{key}: {value}")
# Calling the print_kwargs fucntion
print_kwargs(name="Anand", age=24, city="Hyderabad")
```

```python
# 
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")

print_details(name="Anand", age=30, city="Erode", state="Tamil Nadu", country="India")
```

```python
# Postional Arguments + Keyword Arguments
def print_details(*args , **kwargs): # *args are the postional arguments and **kwargs are the keyword arguments 
    for val in args:
        print(f"postional arguments : {val}")

    for key, value in kwargs.items():
        print(f"keyword arguments : {key} : {value}")


# Calling the print_details function
print_details(1,2,3,4,5,6,7,8,9,0,1,name="Anand", age=30, city="Erode", state="Tamil Nadu", country="India")

```
