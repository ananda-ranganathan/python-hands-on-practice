```python
# Filter Functions in Python
# Defintion: Filter functions are used to filter out elements from an iterable based on some condition.

# Syntax: filter(function, iterable)  
      
# Example: filter(lambda x: x % 2 == 0, range(10))
# Output: <filter object at 0x7f8b8b0c9f60> # iterator object

# Example: list(filter(lambda x: x % 2 == 0, range(10)))
# Output: [0, 2, 4, 6, 8]
```

```python
def even (num):
    if num %2==0:
        return True
    
print(even(24))

# Filter Fucntion with List using the above same function "even" and logic
LIST = [1,1,2,2,3,4,5,56,6,7]
list(filter(even,LIST))
```

```python
# Filter with lambda Fucntions
numbers= [1,2,3,4,5,7,33,83,9]
greater_than_five = list(filter(lambda L : L > 5, numbers))
print(greater_than_five)
```

```python
# Filter function with lambda fucntion with Multiple Conditions
numbers = [1,2,3,4,5,6,7,7,8]
even_and_greater_than_five = list(filter(lambda x : x > 5 and x %2==0, numbers))
print(even_and_greater_than_five)
```

```python
# Filter to check oif the age is greater than 25 in dictionaries

# List of dictionaries 
people =[
    {'name':"Cita",'age': 40},
    {'name':"Jack",'age': 30},
    {'name':"Arun",'age':18}
]

def age_greater_than_25(person):
    return person['age']>25

list(filter(age_greater_than_25,people))

# print(age_greater_than_25)
```

```python
# What is filter() in Python?
# filter() is a built-in function in Python.

# It is used to filter elements from an iterable (like list, tuple, set) based on a condition defined in a function.

# It returns a filter object (an iterator) containing only the elements that satisfy the condition.

# 🔹 Syntax
# python
# filter(function, iterable)
# function → A function that returns True or False for each element.

# iterable → A sequence (list, tuple, set, string, etc.) whose elements you want to filter.

# 🔹 Real-time Short Example (Word-by-Word Explanation)
# python
# def is_even(num):       # Step 1: Define a function
#     return num % 2 == 0 # Step 2: Condition (True if even)

# numbers = [1, 2, 3, 4, 5, 6]   # Step 3: Iterable
# result = filter(is_even, numbers) # Step 4: Apply filter

# print(list(result))     # Step 5: Convert to list to see output
# Trace Explanation:
# def is_even(num): → defines a function that checks if a number is even.

# numbers = [1,2,3,4,5,6] → iterable list.

# filter(is_even, numbers) → applies is_even to each element:

# is_even(1) → False → excluded

# is_even(2) → True → included

# is_even(3) → False → excluded

# is_even(4) → True → included

# is_even(5) → False → excluded

# is_even(6) → True → included

# Output → [2, 4, 6]

# 🔹 How Filter Object is Stored and Where?
# filter() returns a filter object, which is an iterator.

# It doesn’t immediately create a list; instead, it stores a lazy evaluation (elements are computed only when needed).

# To see it visually, you must convert it into a list, tuple, or set:

# python
# print(list(result))   # [2, 4, 6]
# print(tuple(result))  # (2, 4, 6)
# 🔹 Why is filter() Used?
# To extract elements that meet a condition.

# Saves memory because it doesn’t create a new list until you explicitly convert it.

# Useful in data cleaning, preprocessing, and functional programming.

# 🔹 When Does It Take Effect?
# When you iterate over the filter object (e.g., using list(), tuple(), for loop).

# Until then, it’s just a lazy iterator.

# 🔹 Features of filter() Function
# Functional programming style (works with functions like map, reduce).

# Lazy evaluation (efficient memory usage).

# Works with any iterable (list, tuple, set, string).

# Returns only elements that satisfy the condition.

# 🔹 Parameters in filter()
# Function: A function that returns True or False.

# Can be a named function or a lambda function.

# Iterable: Any sequence (list, tuple, set, string, dictionary keys).

# Example with lambda:

# python
# numbers = [10, 15, 20, 25]
# result = filter(lambda x: x > 15, numbers)
# print(list(result))   # [20, 25]
# 🔹 Inbuilt Functions with filter()
# You can use Python’s built-in functions directly inside filter().

# Examples:
# Using str.isalpha

# python
# chars = ['a', '1', 'b', '2', 'c']
# result = filter(str.isalpha, chars)
# print(list(result))   # ['a', 'b', 'c']
# Using str.isdigit

# python
# chars = ['a', '1', 'b', '2', 'c']
# result = filter(str.isdigit, chars)
# print(list(result))   # ['1', '2']
# Using None as function

# python
# values = [0, 1, '', 'hello', [], [1,2]]
# result = filter(None, values)
# print(list(result))   # [1, 'hello', [1,2]]
# 👉 Here, None means "filter out all falsy values".

# 🔹 Visual Representation (Trace Table)
# Element	Condition (is_even)	Result
# 1	False	Excluded
# 2	True	Included
# 3	False	Excluded
# 4	True	Included
# 5	False	Excluded
# 6	True	Included
# ✅ So in short:

# filter() = extracts elements based on condition.

# Returns a lazy filter object.

# Needs conversion (list, tuple) to see results.

# Works with functions, lambdas, and built-ins.

# Efficient and interview-friendly.
```
