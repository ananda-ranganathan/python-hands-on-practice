```python
# lambda functions
# Definition: A lambda function is a small anonymous function that can be defined in a single line using the lambda keyword.
# The lambda function takes any number of arguments but can only have one expression.

# Syntax:
# lambda arguments : expression

# Example:
add = lambda x, y: x + y
print(add(5, 3))  # Output: 8

# Step by Step Explaination

# 1. lambda: keyword
# 2. add: variable name
# 3. lambda x, y: x + y: lambda function
# 4. x, y: arguments
# 5. x + y: expression

# In this example, the lambda function add takes two arguments x and y and returns the sum of x and y.
```

```python
def addition(a,b):
    return a+b # this just one expression; this is just one Logic have been written; this is just one operation have written in one line

addition(4,6)
```

```python
# Conveting the above addition function into lambda function
add = lambda a,b: a+b
# add: variable name ; lambda a,b: a+b: lambda function ; here a,b: arguments ; a+b: expression
print(type(addition))
print(addition(4,6)) # addition(4,6) is normal function call; add(4,6) is lambda function call

```

```python
# Example 3
def even(a):
    if a%2==0:
        return(True)
    
print(even(24))

# Converting into Lambda
even_num = lambda a: a%2==0
print(even_num(24))
```

```python
# Example with multiple Parameters
def addition(a,b,z):
    return a+b+z

addition(12,13,15)

# Converting multiple parameters with Lambda 
add_multi_para = lambda a,b,c : a+b+c
print(add_multi_para(1,2,3))
```
