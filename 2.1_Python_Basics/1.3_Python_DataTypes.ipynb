```python
# Datatypes in Python
int_var = 10      # Integer type
float_var = 10.5  # Float type
str_var = "Hello" # String type
bool_var = True   # Boolean type
print(type(int_var))   # <class 'int'>
print(type(float_var)) # <class 'float'>
print(type(str_var))   # <class 'str'>
print(type(bool_var))  # <class 'bool'> 

# Common datatypes explained
# Integer: Whole numbers without a decimal point (e.g., -3, 0, 42).
# Float: Numbers with a decimal point (e.g., -2.5, 0        .0, 3.14).
# String: Sequence of characters enclosed in quotes (e.g., "Hello", 'Python').
# Boolean: Represents True or False values.

# Datatype definitions
# Data types in Python define the kind of data a variable can hold.
#Important datatypes include:
# They prevent errors and help optimize memory usage.
# They help python interpreter to allocate memory efficiently.
# They make code more readable and maintainable.
# They reduce run time errors by enforcing type constraints.
```

```python
#common Datatypes in Python
# int: Integer type for whole numbers.
age = 25
print(age)

# float: Float type for decimal numbers.
height = 5.9
print(height)

# str: String type for text.
name = "anand"
text_id = "abc&6^%$#@!Xyz"
print(name+text_id)

# bool: Boolean type for True/False values.
is_student = True
print(is_student)
print(type(is_student))

type(True)
print(type(True))

type(False)
print(type(False))
```

```python
#boolean Operations and Comparisons
a=10
b=10
print(a is b)
print(a == b)
print(type(a))
print(type(b))
print(type(a==b))
```
