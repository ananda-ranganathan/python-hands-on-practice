```python
#python Variable assignments
a = 10
name = "Anand"
price = 99.99
print(type(a)) 
print(type(name))   
print(type(price))
```

```python
#python type inference
variable = 20
print(type(variable))
variable = "Now I'm a string"
print(type(variable))

#Type inference in Python means Python automatically detects the datatype of a variable 
# based on the value assigned to it.
# You do not tell Python the type — Python figures it out
```

```python
#Python line continousation
total = 1 + 2 + 3 + \
            4 + 5
print(total)

#Line continuation in Python allows a single statement to span multiple lines using parentheses, 
# brackets, or a backslash for better readability.
#⚠️ Not recommended often (error-prone if space after \).
```

```python
#Multiple Statement on Single Line 
x = 5; y = 10; z = x + y
print(x, y, z)

#Python allows multiple statements on a single line using semicolons, 
# but it is discouraged for clean and readable code.
```
