```python
#Python Variables
# A variable in Python is a name that refers to a value stored in memory.
# Python variables are created when a value is assigned, and their data type 
# is determined automatically at runtime.
# Variables can hold different data types like integers, strings, floats, lists, etc.
# Variable names must start with a letter or underscore, followed by letters, digits, or underscores.
# Python is dynamically typed, so you don't need to declare variable types explicitly.
# Variables can be reassigned to values of different types.
# Variables are case sensitive
# Variables name should be descriptive and follow naming conventions for better code readability.
# They must start with a letter (a-z, A-Z) or an underscore (_), followed by letters, digits (0-9), 
# or underscores.
```

```python
#Declaring and Assinging Variables in Python
x = 5               # Integer assignment    
name = "Anand"     # String assignment
price = 99.99      # Float assignment   
is_available = True  # Boolean assignment
items = [1, 2, 3, 4, 5]  # List assignment

#Printing Variable Values
print(x)                # Output: 5
print(name)             # Output: Anand
print(price)            # Output: 99.99
print(is_available)     # Output: True
print(items)           # Output: [1, 2, 3, 4, 5]

```

```python
#Valid and Invalid Variable Names in Python
# Valid variable name
first_name = "Ananda"    # Valid
last_name = "Ranganathan" 

# Invalid: starts with a digit
2nd_place = "Alice"      # Invalid: starts with a digit 
name="John"          # Valid
Name="Doe"           # Valid: case-sensitive
Name==name   # Invalid: uses comparison operator ==
total$amount = 100       # Invalid: contains special character $
user-name = "Bob"        # Invalid: contains special character -
class = "Math"           # Invalid: uses reserved keyword
# Python variable names cannot start with a digit, contain special characters (except _),
# or use reserved keywords.
# Valid variable names should start with a letter or underscore and can contain letters, digits, and underscores.
# Variable names are case-sensitive.
```

```python
#type checking and conversion in Python
# Type Checking
age = 25
print(type(age))  # Output: <class 'int'>

# Type Conversion
age_str = str(age)
print(type(age_str))  # Output: <class 'str'>

# Converting to different types
height = 5.9
height_int = int(height)
print(height_int)  # Output: 5

# Converting string to integer
num_str = "123"
num_int = int(num_str)
print(num_int)  # Output: 123

#Converting string to integer
Name = "Anand"
print(int(Name))  # Raises ValueError: invalid literal for int() with base 10: 'Anand'
```

```python
#Dynamic Typing in Python
variable = 10       # Initially an integer
print(variable,type(variable))  # Output: <class 'int'>

variable = "Hello"  # Now a string
print(variable,type(variable))  # Output: <class 'str'>

variable = 3.14    # Now a float
print(variable,type(variable))  # Output: <class 'float'>

variable = [1, 2, 3]  # Now a list
print(variable,type(variable))  # Output: <class 'list'>


# Dynamic typing means that the type of a variable is determined at 
# runtime based on the value assigned to it. You can reassign variables to 
# different data types without explicit type declarations.

```

```python
#input and type casting in Python
# Taking user input 
age = input("Enter your age: ")
print("Age:", type(age))  # Output: <class 'str'>
# By default, input() returns a string. To use the input as a different type,
# you need to cast it explicitly.

# Type casting
age = int(input("enter the age: "))  # Casting input to integer
print(age,"Age as integer:", type(age))  # Output: <class 'int'>
```
