```python
# Operators in Python 
# Definition: Operators are special symbols or keywords in Python that perform operations on 
# one or more operands (values or variables) and produce a result.
# They are used to manipulate data and variables in various ways, such as arithmetic calculations, 
# comparisons, logical operations, and more.

# Types of Operators in Python:
# 1. Arithmetic Operators
# 2. Comparison Operators
# 3. Logical Operators
# 4. Assignment Operators
# 5. Bitwise Operators
# 6. Membership Operators
# 7. Identity Operators
# 8. Special Operators (like the Ternary Operator)
```

```python
# Arithmetic Operators 
# These operators are used to perform mathematical operations between numeric values.
# Addition (+), Subtraction (-), Multiplication (*), Division (/), Modulus (%), 
# Exponentiation (**), Floor Division (//)

a = 10
b = 5
# Addition operator adds two operands
print("Addition:", a + b)          # Output: 15
# Subtraction operator subtracts the second operand from the first
print("Subtraction:", a - b)       # Output: 5
# Multiplication operator multiplies two operands
print("Multiplication:", a * b)    # Output: 50
# Division operator returns a float result
print("Division:", a / b)          # Output: 2.0
# Modulus operator returns the remainder of the division
print("Modulus:", a % b)           # Output: 0
# Exponentiation operator raises the first operand to the power of the second
print("Exponentiation:", a ** b)   # Output: 100000
# Floor Division operator returns the largest integer less than or equal to the division result
print("Floor Division:", a // b)    # Output: 2
```

```python
# Comparison Operators
# Definition: Comparison operators are used to compare two values or expressions 
# and return a Boolean result (True or False).
# They help in making decisions based on conditions in the code.

# Types of Comparison Operators:
# Equal to (==), Not equal to (!=), Greater than (>), Less than (<), 
# Greater than or equal to (>=), Less than or equal to (<=)

# Example Usage of Comparison Operators:
x = 10
y = 20
# Boolean Comparisons x is y returns True if both operands refer to the same object
print(x is y)                           # Output: False
# Equal to (==) returns True if both operands are equal
print("Equal to:", x == y)                # Output: False
# Not equal to (!=) returns True if operands are not equal
print("Not equal to:", x != y)            # Output: True
# Greater than (>) returns True if the left operand is greater than the right
print("Greater than:", x > y)             # Output: False
# Less than (<) returns True if the left operand is less than the right
print("Less than:", x < y)                # Output: True
# Greater than or equal to (>=) returns True if the left operand is greater than or equal to the right
print("Greater than or equal to:", x >= y) # Output: False
# Less than or equal to (<=) returns True if the left operand is less than or equal to the right
print("Less than or equal to:", x <= y)   # Output: True

a = 45
b = 45
print (a<=b)

```

```python

# Logical Operators
# Definition: Logical operators are used to combine multiple Boolean expressions
# and return a single Boolean result (True or False).
# They are commonly used in conditional statements and loops to control the flow of execution.

# Types of Logical Operators:
# AND (and), OR (or), NOT (not)

# Example Usage of Logical Operators:
x = 10
y = 20
# AND operator returns True if both operands are True
print("AND:", (x < y) and (x > 5))      # Output: True
# OR operator returns True if at least one operand is True
print("OR:", (x > y) or (x > 5))       # Output: True
# NOT operator returns the opposite Boolean value of the operand
print("NOT:", not(x < y))               # Output: False

x=True
y="false"
result = not y
print (result)

```

```python
# Assignment Operators
# Definition: Assignment operators are used to assign values to variables.
# They provide a shorthand way to perform operations on variables and update their values.

# Types of Assignment Operators:
# Simple assignment (=), Add and assign (+=), Subtract and assign (-=),
# Multiply and assign (*=), Divide and assign (/=), Modulus and assign (%=),
# Exponentiation and assign (**=), Floor division and assign (//=)

# Example Usage of Assignment Operators:
a = 10
b = 5
# Simple assignment operator (=) assigns the value of the right operand to the left operand
c = a + b  # c = 10 + 5
#step by step calculation:
# 1. a + b = 10 + 5 = 15
print("Value of c after simple assignment:", c)  # Output: 15

# Add and assign operator (+=) adds the right operand to the left operand and assigns the result to the left operand
a += b  # a = a + b
# step by step calculation:
# 1. a + b = 10 + 5 = 15
print("Value of a after a += b:", a)  # Output: 15

# Subtract and assign operator (-=) subtracts the right operand from the left operand and assigns the result to the left operand
a=15
b=5
a -= b  # a = a - b
# step by step calculation:
# 1. a - b = 15 - 5 = 10
print("Value of a after a -= b:", a)  # Output: 10

# Multiply and assign operator (*=) multiplies the left operand by the right operand and assigns the result to the left operand
a=10
b=5
a *= b  # a = a * b 
# step by step calculation:
# 1. a * b = 10 * 5 = 50
print("Value of a after a *= b:", a)  # Output: 50

# Divide and assign operator (/=) divides the left operand by the right operand and assigns the result to the left operand
a=50
b=5
a /= b  # a = a / b
# step by step calculation:
# 1. a / b = 50 / 5 = 10.0
print("Value of a after a /= b:", a)  # Output: 10.0

# Modulus and assign operator (%=) returns the remainder of the division of the left operand by the right operand
x = 4
y = 3
x %= y  # x = x % y

# Step by step calculation:
# 1. 4 / 3 = 1.333...
# 2. 1 * 3 = 3
# 3. 4 - 3 = 1
print("Value of x after x %= y:", x)  # Output: 1
#Explanation: 4 % 3 = 1 (4 divided by 3 leaves a remainder of 1)

# Exponentiation and assign operator (**=) raises the left operand to the power of the right operand
x = 2
y = 5
x **= y  # x = x ** y 
# Step by step calculation:
# 1 * 2 = 2
# 2 * 2 = 4
# 4 * 2 = 8
# 8 * 2 = 16    
# 16 * 2 = 32       
print("Value of x after x **= y:", x)  # Output: 32
#Explanation: 2 ** 5 = 32 (2 raised to the power of 5 is 32)


# Floor division and assign operator (//=) divides the left operand by the right operand
# and assigns the largest integer less than or equal to the result
x = 2
y = 3
x //= y  # x = x // y

# Step by step calculation:
# 1. 2 / 3 = 0.666...
# 2. The largest integer less than or equal to 0.666... is 0
print("Value of x after x //= y:", x)  # Output: 0
# Explanation: 2 // 3 = 0 (The largest integer less than or equal to 0.666... is 0)
```

```python
# Bitwise Operators
# Definition: Bitwise operators are used to perform operations on individual bits of integer values.
# They treat the operands as sequences of binary digits (bits) and perform operations on each corresponding bit.
# Bitwise operators are commonly used in low-level programming, data manipulation, and optimization tasks.

# Types of Bitwise Operators:
# AND (&), OR (|), XOR (^), NOT (~), Left Shift (<<), Right Shift (>>)

# Example Usage of Bitwise Operators:
a = 10  # In binary: 1010
b = 4   # In binary: 0100
# AND operator (&) performs a bitwise AND operation on each pair of corresponding bits
#step by step calculation:
# 1. 1010
# 2. 0100
# 3. -----
# 4. 0000
print("Bitwise AND:", a & b)          # Output: 0 (In binary: 0000)
#Explanation:
# Bitwise AND compares each bit of two numbers and returns 1 if both bits are 1, otherwise returns 0.
# Example: 1010 & 0100 = 0000 (10 & 4 = 0) 

# OR operator (|) performs a bitwise OR operation on each pair of corresponding bits
a = 10  # In binary: 1010
b = 4   # In binary: 0100
#step by step calculation:
# 1. 1010
# 2. 0100
# 3. -----
# 4. 1110
print("Bitwise OR:", a | b)           # Output: 14 (In binary: 1110)
#Explanation:
# Bitwise OR compares each bit of two numbers and returns 1 if at least one bit is 1, otherwise returns 0.
# Example: 1010 | 0100 = 1110 (10 | 4 = 14)

# XOR operator (^) performs a bitwise XOR operation on each pair of corresponding bits
a = 25   # In binary: 11001
b = 4   # In binary: 00100
#step by step calculation:
# 1. 11001
# 2. 00100
# 3. -----  
# 4. 11101  
print("Bitwise XOR:", a ^ b)          # Output: 29 (In binary: 11101)
#Explanation:
# Bitwise XOR compares each bit of two numbers and returns 1 if the bits are different, otherwise returns 0.
# Example: 11001 ^ 00100 = 11101 (25 ^ 4 = 29)

# NOT operator (~) performs a bitwise NOT operation, flipping each bit of the number
a = 10  # In binary: 0000...1010
#step by step calculation:
# 1. 0000...1010
# 2. -----
# 3. ...11110101
print("Bitwise NOT:", ~a)              # Output: -11 (In binary: ...11110101)
#Explanation:
# Bitwise NOT flips each bit of the number, changing 1s to 0s and 0s to 1s.
# Example: ~1010 = ...11110101 (~10 = -11 in two's  complement representation)

# Left Shift operator (<<) shifts the bits of the first operand to the left by the number of positions
# specified by the second operand
a = 10  # In binary: 1010
#step by step calculation:
# 1. 1010
# 2. Shift left by 2 positions: 101000  
print("Left Shift:", a << 2)          # Output: 40 (In binary: 101000)
#Explanation:
# Left Shift shifts the bits of the number to the left by the specified number of positions,
# filling the rightmost bits with 0s.

# Right Shift operator shifts the bits of the first operand to the right by the number of positions 
# specified by the second operand
a = 10  # In binary: 1010
#step by step calculation:
# 1. 1010
# 2. Shift right by 2 positions: 0010
print("Right Shift:", a >> 2)         # Output: 2 (In binary: 0010) 
#Explanation:
# Right Shift shifts the bits of the number to the right by the specified number of positions,
# filling the leftmost bits with 0s (for positive numbers) or 1s (for negative numbers).
```

```python
# Membership Operators
# Definition: Membership operators (in, not in) are used to test whether a value or variable is present in a sequence
# (such as a string, list, tuple, set, or dictionary) or not.
# They return a Boolean result (True or False) based on the presence of the value in the sequence.

# Types of Membership Operators:
# in, not in    

# Example Usage of Membership Operators:
sequence = [1, 2, 3, 4, 5]
# 'in' operator checks if a value is present in the sequence
print(3 in sequence)        # Output: True

# 'not in' operator checks if a value is not present in the sequence
print(6 not in sequence)    # Output: True
```

```python
# Identity Operators
# Definition: Identity operators (is, is not) are used to compare the memory locations of two objects.
# They return True if the operands refer to the same object in memory, otherwise they return False.

# Types of Identity Operators:
# is, is not

# Example Usage of Identity Operators:
a=10
b=10

# 'is' operator checks if both operands refer to the same object
print(a is b)           # Output: True

# 'is not' operator checks if both operands do not refer to the same object
print(a is not b)      # Output: False  
```

```python
# Special Operators (like the Ternary Operator)
# Definition: Special operators (like the Ternary Operator) are unique operators that do not fit 
# into the standard categories of operators.
# One common example is the Ternary Operator, which allows for conditional expressions in a single line.

# The Ternary Operator has the following syntax:
# value_if_true if condition else value_if_false

# Example Usage of Ternary Operator:
age = 18

# Using the Ternary Operator (value_if_true if condition else value_if_false) to determine if a person is an adult or a minor
status = "Adult" if age >= 18 else "Minor"
print("Status:", status)  # Output: Adult
# Explanation: The Ternary Operator evaluates the condition (age >= 18). If the condition is True, it returns "Adult"; 
# otherwise, it returns "Minor".
```
