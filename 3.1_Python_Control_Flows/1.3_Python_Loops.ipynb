```python
# Loops in Control Flow
# Loops in Python are used to iterate over a block of code as many times as needed.
# There are two types of loops in Python: for loop and while loop.

# For Loop
# A for loop is used to iterate over a sequence (list, tuple, string) or other iterable object.
# It is used to execute a block of code for each item in the sequence.

# While Loop
# A while loop is used to execute a block of code as long as a certain condition is true.
# It is used to execute a block of code until a certain condition is met.

# Nested Loops
# A nested loop is a loop inside another loop.
# It is used to execute a block of code for each item in one sequence and for each item in another sequence.

# Loop Control Statements
# Loop control statements are used to modify the behavior of loops.
# The break statement exits the loop prematurely, 
# The continue statement skips the current iteration and moves to the next iteration.
# The pass statement is a placeholder that does nothing. 

# Iteration
# Iteration is the process of repeatedly executing a block of code until a certain condition is met.
# In Python, iteration is implemented using loops.

# Iterators
# An iterator is an object that allows you to iterate over a sequence of elements.
# It is implemented using the __iter__() and __next__() methods.
# The __iter__() method returns the iterator object, and the __next__() method returns the next element in the sequence.

# Generators
# A generator is a function that returns an iterator.
# It is implemented using the yield keyword.
# A generator is a function that returns an iterator, and it is implemented using the yield keyword.

# Generator Expressions
# A generator expression is a short way to create a generator object.
# It is implemented using the (expression) syntax.
# A generator expression is a short way to create a generator object, and it is implemented using the (expression) syntax.

# Range
# A range is a sequence of numbers.
# It is implemented using the range() function.
# A range is a sequence of numbers, and it is implemented using the range() function.
```

```python
# While Loop
# While loop runs until the specific conditions met TRUE
#Examples 
count = 0
while count < 5:
        count = count + 1
        print(count)

# Break , Continue , 

# Pass
# The pass statement is a null operation; it does nothing.

for i in range(5):
    if i==3:
        pass
    print(i)

# Nested Loop
# In nested loop print statement use f string formatting

# a loop inside a loop

for i in range(3):
    for j in range(2):
        print(f"i:{i} and j:{j}")
```

```python
# Square
num = [1,2,3,4,5,6,7,8]
def square(num):
    result = []
    for nums in num:
        result.append(nums**2)
    return result

print(square(num))

# Imagine a bucket 🪣
# result = [] → empty bucket
# append() → drop one stone inside
# Loop → keeps dropping stones
# return result → show full bucket
# Without bucket → stones fall on ground 😄
```
