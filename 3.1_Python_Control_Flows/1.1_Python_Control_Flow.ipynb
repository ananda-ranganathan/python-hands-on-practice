```python
# Control Flow in Python
# Definition: Control flow in Python refers to the order in which individual statements, 
# instructions, or function calls are executed or evaluated.
# It determines how the program's execution proceeds based on certain conditions or loops.
# Control flow is managed using conditional statements (if, elif, else) and loops (for, while).

# Types of Control Flow Statements:
# 1. Conditional Statements
# 2. Loops

# Conditional Statements
# Conditional statements allow you to execute certain blocks of code based on specific conditions.

# Types of Conditional Statements:
# 1. if statement
# Definition: An if statement is used to test a specific condition. If the condition evaluates to True,
# the block of code within the if statement is executed.

# 2. if-else statement
# Definition: An if-else statement is used to test a condition and provide an alternative block of code
# to execute if the condition evaluates to False.

# 3. if-elif-else statement
# Definition: An if-elif-else statement allows you to test multiple conditions sequentially.
# If one of the conditions evaluates to True, the corresponding block of code is executed.
# If none of the conditions are True, the else block is executed.

# 4. nested if statements
# Definition: Nested if statements are if statements placed inside another if statement.
# This allows for more complex decision-making
# by testing additional conditions within the scope of an outer if statement.

# 5. switch-case statement (using match-case in Python 3.10+)
# Definition: A switch-case statement (implemented using match-case in Python 3.10 and later)
# allows you to execute different blocks of code based on the value of a variable or expression.

# Loops
# Loops allow you to execute a block of code multiple times based on certain conditions.

# Types of Loops:
# 1. for loop  
# Definition: A for loop is used to iterate over a sequence (such as a list, tuple, string, or range)
# of values, executing a block of code for each item in the sequence.
#  
# 2. while loop
# Definition: A while loop repeatedly executes a block of code as long as a specified condition is True.

# 3. nested loops
# Definition: Nested loops are loops placed inside another loop.
# This allows for more complex iterations, such as iterating over multi-dimensional data structures.

# 4. loop control statements (break, continue, pass)
# Definition: Loop control statements are used to modify the behavior of loops.
# The break statement exits the loop prematurely, the continue statement skips the current iteration
# and moves to the next iteration, and the pass statement is a placeholder that does nothing.
```
