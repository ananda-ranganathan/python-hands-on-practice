```python
# Conditional Statements
# Conditional statements allow you to execute certain blocks of code based on specific conditions.

# Types of Conditional Statements:
# 1. if statement
# Definition: An if statement is used to test a specific condition. If the condition evaluates to True,
# the block of code within the if statement is executed.

# 2. if-else statement
# Definition: An if-else statement is used to test a condition and provide an alternative block of code to execute if the condition evaluates to False.
# above definition means if condition is false then else block will be executed

# 3. if-elif-else statement
# Definition: An if-elif-else statement allows you to test multiple conditions sequentially.
# If one of the conditions evaluates to True, the corresponding block of code is executed.
# If none of the conditions are True, the else block is executed.

# 4. nested if statements
# Definition: Nested if statements are if statements placed inside another if statement.
# This allows for more complex decision-making
# by testing additional conditions within the scope of an outer if statement.

# Outdated
# 5. switch-case statement (using match-case in Python 3.10+)
# Definition: A switch-case statement (implemented using match-case in Python 3.10 and later)
# allows you to execute different blocks of code based on the value of a variable or expression.
```

```python
# 1. if statement
# Definition: An if statement is used to test a specific condition. If the condition evaluates to True,
# the block of code within the if statement is executed.

# Syntax:
# if condition:
#     block of code to execute if condition is True

# Example:
Number = 10
if Number <= 10 and Number > 1:
    print("The number is between 1 and 10")
```

```python
#if-else Statement 
# used for decision making 
# else block will run only when if block have been False 
# Example 
age = 15
if age > 15 and age < 18:
    print("I am not eligibe to Vote")
else:
    print("i am eligible for first time Voter")
```

```python
#if-elif-else statement
#Used when we want to check multiple conditions in sequence
#the if block runs when first if condition met true
#if not, python will move on to elif multiple elif block if the first if is False, 
# when the elif block condition met true, the corresponding elif block will be printed 
#if none of the if and elif block is true and says the condition is false 
#the else block will get printed
#Python does not execute if any other below's block, when the elif or else condiion is True ONCE

#example 
marks = 85
if marks != 85:
    print("grade f")
elif marks >= 95:
    print("Grade A")
elif marks >= 75:
    print("grade b")
elif marks >= 60:
    print("Grade C")
else:
    print("Pass")


# seperate Blocks
if marks == 100:
     print("Extra if")
else:
    print("Pass passed")
```

```python
#Nested If Statement 
# If within If statement 
# For complex conditions and multilple level of conditions = Like a decision tree

#Example
num = 16
if num == 16:
    print("Positive Number")
    if num %2 == 0:
        print("Even Number") # this block will prints if, IF outer block will be True
        if num - 2 == 18:
            print("16-2=14")
    else:
        print("odd number") # this block will prints when ALL Inner IF block Met False 
else:
    print("Zero or Positive Number") # only prints this outer else block if outer First IF block gets False
```
