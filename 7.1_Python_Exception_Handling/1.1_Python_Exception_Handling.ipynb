```python
# Exception Handling in Python
# Definition: Exception handling is a way to handle the errors in a program.
# It is used to catch and handle the exceptions that occur in a program.

# There are two types of exceptions in Python: built-in exceptions and user-defined exceptions.
# Built-in exceptions are exceptions that are raised by the Python interpreter.
# User-defined exceptions are exceptions that are raised by the user.
```

```python
# try and except block and calling child class to print the message 
try:
    a=2
    b=0 # this will raise a ZeroDivisionError exception: 
    print(a/b)
except ZeroDivisionError as ex: # ZeroDivisionError is a child class of Exception
    print(ex)

# Here, ZeroDivisionError: division by zero is caught by the except block
```

```python
# Try and Except block with out calling NameError exception

try:
    a = no_variable
except:
    print("The Variable has not been Assigned yet...")

# Here, NameError: name 'no_variable' is not defined is not caught by the except block
# Printing message in except block is good pratice and useful for end users
```

```python
# Try and Except block with calling NameError exception

try:
    a = no_variable
except NameError as ex:
    print(ex)

# Here, NameError: name 'no_variable' is not defined is not caught by the except block
# Printing message in except block is good pratice and useful for end users
# Same message is print by name error class exception
```

```python
# Print message include child exception class
try:
    result=1/0
except ZeroDivisionError as ex:
    print(ex)
    print("Please enter the denominator greater than Zero")
```

```python
# try and except block and calling base Parent Exception class which holds all inbuild childs exceptions class
try:
    result = 1/3  # 1/3 = 0.3333333333333333
    a=invalid # this will raise a NameError exception:
except ZeroDivisionError as ex: # ZeroDivisionError is a child class of Exception
    print(ex)
    print("Please enter the denominator as greater than Zero")
except Exception as Ex1: # Exception is a base class (Parent Class) for all exceptions
    print(Ex1)
    print("Main Exception class got caught here")

# Here, ZeroDivisionError: division by zero is not caught by the except block because its divisible 1/3 = 0.33333
# NameError: name 'invalid' is not defined is CAUGHT BY THE PARENT CLASS Exception
# Note: the parent base Exception class should not be before the child class Exception
```

```python
# Try Except by providing user input
try:
    num = int(input("Enter a number: "))
    result = 10/num
except ValueError:
    print("Invalid input. Please enter a valid number.")
except ZeroDivisionError:
    print("Cannot divide by zero.")

# Note: provide input as Zero and run ; Provide string chars and run
```

```python
# Try , Except and Else Block 

try:
    num = 1/2
except ZeroDivisionError as exp:
    print(exp,"Please enter the denominator as greater than Zero")
else:
    print("No Exception - else block running")

# num = 1/0 will raise ZeroDivisionError exception and except block will run

# num = 1/2 will not raise ZeroDivisionError exception and else block will run

```

```python
#  Try and Multiple Except blocks with else block 

try:
    num = int(input("Enter a number: "))
    result = 10/num
except ValueError:
    print("Invalid input. Please enter a valid number.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
except Exception as EXP:
    print(EXP)
else:
    print(f"the result is {result} ")

# Note: 
# provide input as Zero and run ; Provide string chars and run
# num = 1/0 will raise ZeroDivisionError exception and except block will run
# num = 10/"Anand" will raise ValueError exception and except block will run
# num = 1/2 will not raise ZeroDivisionError exception or ValueError exception
#  and else block will run with printing result

```

```python
#  Try and Multiple Except blocks else Block with finally block

try:
    num = int(input("Enter a number: "))
    result = 10/num
except ValueError:
    print("Invalid input. Please enter a valid number.")
except ZeroDivisionError:
    print("Cannot divide by zero.")
except Exception as EXP1:
    print(EXP1)
else:
    print(f"the result is {result} ")
finally:
    print("finally block running")

# Note: 
# provide input as Zero and run ; Provide string chars and run
# num = 1/0 will raise ZeroDivisionError exception and except block will run
# num = 10/"Anand" will raise ValueError exception and except block will run
# num = 1/2 will not raise ZeroDivisionError exception or ValueError exception
#  and else block will run with printing result
# Finnally block will always run whether the case is handled or not
```
