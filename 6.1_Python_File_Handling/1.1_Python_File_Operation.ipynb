```python
# File Operations - Read and Write Files
# Definition: File operations allow you to perform various operations on files, 
# such as reading, writing, appending, and deleting them.\n
# They are a fundamental part of file management and data processing in Python.\n

# Read - The read() method reads the entire contents of a file and returns it as a string.
# Write - The write() method writes a string to a file.\n
# Append - The append() method appends a string to the end of a file.\n
# Delete - The delete() method deletes a file.\n
```

```python
# Read the whole file
with open('read_file_example.txt','r') as read_file:
    read_content = read_file.read()
    print(read_content)
```

```python
# Read the whole file
with open('read_file_example.txt','r') as read_file:
    read_content = read_file.read()
    print(read_content)

```

```python
# read file Line by Line
with open('read_file_example.txt','r') as read_file: 
    for line in read_file:
        print(line.strip())  
        # strip() method is used to remove leading and trailing whitespace charactersS,
        # Remove such as spaces, tabs, and newlines, from a string.
```

```python
# Write into the same .txt file ((read_file_example.txt))
with open('read_file_example.txt','w') as write_file:
    write_file.write("Hello am writing something\n") # Be Careful it will overwrite the same file content 
    # everything will be erased and overwritten with the new content


    
```

```python
# Write to the same txt.file name without overwriting
with open('read_file_example.txt','a') as write_file:
    write_file.write("Hello am writing something new and not overwritten\n")
```

```python
# Writing a List of Lines to a same .txt file
lines_lst= ['this is first line of sentence \n','this is second line of sentence \n','this is third line of sentence \n']
with open ('read_file_example.txt','a') as write_file:
    write_file.writelines(lines_lst)
```

```python
# Write and Read - Binary Files
binary_data = b'\x01\x02\x03\x04\x05'
with open('binary_file.bin', 'wb') as binary_file:
    binary_file.write(binary_data)

with open('binary_file.bin', 'rb') as binary_file:
    read_binary_data = binary_file.read()
    print(read_binary_data)
```
