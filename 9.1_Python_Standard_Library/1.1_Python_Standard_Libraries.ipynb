```python
# Data Serialization # Serialization is the process of converting a data structure or object into a format 
# that can be easily stored and later retrieved.   

import json # json is a data interchange format based on a subset of the JavaScript Object Notation (JSON) standard 
data={'name':"Anand",'Age':29}

# json_str=json.dumps(data, indent=4) - json.dumps(data, indent=4, sort_keys=True) 
json_str=json.dumps(data) # dumps - convert python object into json string

print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str) # loads - convert json string into python object
print(parsed_data)
print(type(parsed_data))

# import pickle
# data = {'a': [1, 2.0, 3, 4+6j], 'b': ("character string", b"byte string"), 'c': {None, True, False}}
# pickle.dump(data, open("data.p", "wb"))
# print(pickle.load(open("data.p", "rb")))
```

```python
# Date Time Delta Library
from datetime import datetime,timedelta
live_time = datetime.now()
print(f"Live time is {live_time}") # print(live_time)

#Yeterday Date and time form Live time
yesterday = live_time - timedelta(days=1)  # Yesterday - use minus sign
print(f"Yesterday is {yesterday}") # print(yesterday)
# Tomorrow Date and time form Live time
tomorrow = live_time + timedelta(days=1)  # Tomorrow - use plus sign
print(f"Tomorrow is {tomorrow}") # print(tomorrow)
# import datetime
# from datetime import date
# from datetime import timedelta
# from datetime import datetime
# from datetime import time
# from datetime import timezone
# from datetime import tzinfo
# from datetime import tzlocal
```
