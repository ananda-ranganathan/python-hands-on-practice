```python
# Python Polymorphism in OOPS
# Definition : Polymorphism is the ability of objects to take on many forms.
# In Python, polymorphism refers to the ability of objects of different classes to respond to the same method name in different ways.
# Polymorphism allows you to write code that can work with objects of different classes, as long as they share a common interface

# Types of Polymorphism in Python
# 1. Method Overloading : Multiple methods with the same name but different parameters are called method overloading 
# 2. Method Overriding : Multiple methods with the same name and same parameters are called method overriding 
# 3. Operator Overloading : Multiple operators with the same name but different parameters are called operator overloading 
# 4. Duck Typing : Python uses duck typing to determine the type of an object based on its behavior rather than its actual type 
```

```python
# Method Overloading
# Simple : with in the class, same function or method name with passing different parameters 
# if we overload any extra parameters just initalize default arugument lets c = 0 

class Demo:
    def add(self,a,b):
        return a+b
    
d=Demo()
print(d.add(2,3))
    
```
