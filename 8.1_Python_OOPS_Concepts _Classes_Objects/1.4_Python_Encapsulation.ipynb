```python
# Encapsulation in Python 
# Encapsulation is a fundamental concept in object-oriented programming (OOP) that means bundling data and methods inside a 
# class while restricting direct access to certain attributes, ensuring data security, data hiding, data protection and abstraction. 
# Encapsulation will hide the internal details of a class from the outside world.
# Encapsulation can be achieved in Python by using access modifiers, such as public, protected, and private.

# Public: Accessible from anywhere
# Private: Accessible only within the class (__)(data hiding)(used by double underscore)
# Protected: Accessible within the class and its subclasses(derived classes) (_)(used by single underscore)
# getter and setter methods
# getter :  A method that retrieves (gets) the value of an attribute.
# setter : A method that updates (sets) the value of an attribute, often with validation or logic.

    # In Python, we usually use the @property decorator for getters and @<property>.setter for setters.
```

```python
# Public Variable in Encapsulation (Public Access Modifiers)
class Person(): # Created the class called Person
    def __init__(self,name,age): # Creating constructor with self,name,age parameters
        self.name=name   #  Public Access Modifier (Public Variable)
        self.age=age     #  Public Access Modifier (Public Variable)

# The above block of code is one class called Person and terminated with the indentation
# Public Variables will allow the variables to Accessible anywhere (inside or outside the class)

# calling the class Person (public Variables(name, age) ) outside the class
# below the code starts new (outside the class Person )

pub_person = Person("Anand",20) # passing the arguments(values) to the public variables(name, age)
print(pub_person.name) # printing the statement by calling object created 'pub.person' . (variable'name')
dir(pub_person) # dir helps to give output of the methods used (__init__) and also along with variables/para that we created at last line
```

```python
# Private Variable in Encapsulation (private Access Modifiers)
class Person(): # Created the class called Person
    def __init__(self,name,age,gender): # Creating constructor with self,name,age,gender parameters
        self.name=name   #  public Access Modifier (public Variable)
        self.__age=age     # Private Access Modifier (private Variable)(age denoted by double underscore[__])
        self.__gender=gender # Private Access Modifier (Private Variable)(gender denoted by double underscore[__])

# The above block of code is one class called Person and terminated with the indentation
# Public Variables will allow the variables to Accessible anywhere (inside or outside the class)
# BUT PRIVATE VARIABLES DOES NOT ALLOW to ACCESSIBLE OUTSIDE THE CLASS 
# (EVEN SUB/DERIVED CLASS NOT ACCESSIBLE OUTSIDE), means it only accessible within the sameclass using getter and setter methods

# calling the class Person (public Variables(name), private Variable (age,gender)) outside the class
# below the code starts new (outside the class Person )

pri_person = Person("Anand",20,"Male") # passing the arguments(values) to the public variables(name, age,gender)

print(pri_person.name) # printing the statement by calling object created 'pri.person' . (public variable'name')
print(pri_person.gender) # printing the statement by calling object created 'pri.person'(private variable'name') 
#(throws Attributes error 'gender' not founf as calling outside person class)
print(pri_person.age) # printing the statement by calling object created 'pri.person' . (private variable'name')
#(throws Attributes error 'age' not found as calling outside person class)
dir(pri_person) # dir helps to give output of the methods used (__init__) and also along with variables/para that we created at last line
# If we notice in dir output, 
# denoting _Person__age, _Person__gender (appended to the Person Class not outside) is private access modifiers 
# while at last name variable is denoted as publicly 
```

```python
# Protected Variable in Encapsulation (protected Access Modifiers) : Can be accessed within the class and its subclasses but not outside the class
class Student():
    def __init__(self,roll_no,name,course,marks,gender): # Creating constructor with self,roll_no,name,course,marks,gender parameters
        self.roll_no=roll_no   # public Access Modifier (public Variable)
        self.name=name     # public Access Modifier (public Variable)
        self.__couse=course  # Private Access Modifier (private Variable)(course denoted by double underscore[__])
        self._marks=marks      # Protected Access Modifier (protected Variable)(marks denoted by single underscore[_])
        self._gender=gender # Protected Access Modifier (protected Variable)(gender denoted by single underscore[_])

# stu = Student(2,"Rani","python",23,"Female")
# print(stu._marks) # printing the statement by calling object created 'stu' . (protected variable'marks') -> Accessible within the same class.

# The above block of code is one class called Person and terminated with the indentation
# Public Variables will allow the variables to Accessible anywhere (inside or outside the class)
# PRIVATE VARIABLES DOES NOT ALLOW to ACCESSIBLE OUTSIDE THE CLASS, means it only accessible within the same parent class
# PROTECTED VARIABLES DOES NOT ALLOW to ACCESSIBLE OUTSIDE THE CLASS, means it only accessible within the sameclass and its subclasses(derived classes)

# Creating another NEW Derived or sub or child (graduate)class to access the protected variables 
class Graduate(Student): # Created a child class called Graduate and inheriting the parent class called Student
    def __init__(self,roll_no,name,course,marks,gender):
        super().__init__(roll_no,name,course,marks,gender) # super() is used to access the parent class variables

gradu_stu = Graduate(1,"Anand","Data Science",80,"Male") # passing the arguments(values) to the sub class.

print(gradu_stu.roll_no) # printing the statement by calling object created 'gradu_stu' . (public variable'name')
print(gradu_stu.name) # printing the statement by calling object created 'gradu_stu'(public variable'name' )
print(gradu_stu._marks) # printing the statement by calling object created 'gradu_stu' . (protected variable'_marks')(access in derived class)
print(gradu_stu._gender) # printing the statement by calling object created 'gradu_stu' . (protected variable'_gender')(access in derived class)
#print(gradu_stu.__couse) # printing the statement by calling object created 'gradu_stu'(private variable'name'Throws Attributes error as calling outside person class)
# dir(gradu_stu)

```

```python
# Accessing the private variables from the methods of the same class using getter and setter methods
class College_Student(): # Created the Parent class called College_Student
    def __init__(self,name,semester,cgpa,result): # Creating Constructor with parameters
        self.name=name  # Public access
        self._semester=semester # Protected Access
        self.__cgpa=cgpa  # Private Access 
        self.__result=result # Private Access
# __cgpa (private variable) - I need to probably NOT to display the __cgpa anywhere, it should be restricted within the class itself.. thats my aim.
# In-order to access the particular private variables inside the class, I can do it by generating getter and setter methods.

# Getter and setter method for the __result
    def get_result(self):
        return(self.__result)
# Setter method for the __result
    def set_result(self,result):
        self.__result=result

# Getter method for the __cgpa (used to Retrieving the data)
    def get_cgpa(self): # Created a getter method called get_cgpa; exposing the __cgpa to outside world.
        return (self.__cgpa)
# Setter method for the __cgpa (used to changing the cgpa values)
    def set_cgpa(self,cgpa): # Created method that updates or change any values of an attribute(lets cgpa=7.98 to 8.9) for validations.
        if cgpa > 2.0:
            self.__cgpa = cgpa 
        else:
            print("Invalid cgpa")
# creating object coll_stu and class name and passing argument values
coll_stu = College_Student("Ramu","2nd",4.4,"pass")
## Access and modify the private variables using getter and setter 
print(coll_stu.get_cgpa())
print(coll_stu.get_result())

coll_stu.set_cgpa(6)
print(coll_stu.get_cgpa())

coll_stu.set_cgpa(-3)
```
