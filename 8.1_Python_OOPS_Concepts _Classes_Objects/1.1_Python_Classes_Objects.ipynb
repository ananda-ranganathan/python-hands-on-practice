```python
# OOPS in python
# Definition: Object Oriented Programming (OOP) is a programming paradigm that provides a means of structuring 
# programs so that properties and behaviors are bundled into individual objects.

# Types of OOP:
# 1. Class-based OOP # means that classes are used to create objects 
# 2. Object-based OOP # means that objects are used to create classes

# What is a Class?
# A class is a blueprint or a template for creating objects. It defines the attributes and methods that 
# objects of that class will have.    

# What is an Object?
# An object is an instance of a class. It is a specific realization of a class. 
# It has its own set of attributes and methods.   

# What is an Instance?
# An instance is a specific object created from a class. It is a concrete realization of a class. 
# It has its own set of attributes and methods. 

# What is an Instance Variable?
# An instance variable is a variable that is associated with an instance of a class. It is a variable that is 
# defined inside a class and can be accessed on an object of that class.

# What is an instance Method?
# An instance method is a method that is defined inside a class and can be called on an object of that class.

# What is an Attribute?
# An attribute is a variable that is associated with a class. It is a variable that is defined inside a class and 
# can be accessed on an object of that class.  

# What is a Method?
# A method is a function that is associated with a class. It is a function that is defined inside a class and 
# can be called on an object of that class. 

# What is a Constructor?
# A constructor is a special method that is called when an object is created from a class. It is a function 
# that is defined inside a class and can be called on an object of that class.  
# It is used to initialize the object.
# It is used to set the initial state of the object.
# It is used to create the object.

# super() is a special keyword used to call the constructor of the parent class.   

# What is a Destructor?
# A destructor is a special method that is called when an object is destroyed. It is a function that is 
# defined inside a class and can be called on an object of that class.

# What is a Singleton?
# A singleton is a class that has only one instance. It is a class that is defined inside a class and can 
# be called on an object of that class.

# What is a Static Method?
# A static method is a method that is defined inside a class and can be called on an object of that class.

# What is a Class Method?
# A class method is a method that is defined inside a class and can be called on a class.

# What is a Property?
# A property is a variable that is associated with a class. It is a variable that is defined inside a class and 
# can be accessed on an object of that class.

# What is a Data Class?
# A data class is a class that is defined inside a class and can be called on an object of that class.

# What is a Magic Method?
# A magic method is a method that is defined inside a class and can be called on an object of that class.
```

```python
# Class means blueprint or template for creating objects
class Car:  
    pass  # pass means do nothing
audi = Car() 
bmw = Car()   
# class Car: defines a blueprint, and audi = Car() and bmw = Car() create two separate objects (instances) of that class
# A class is a template, and audi and bmw are real objects created from that template.
### class Car: 
# class → Python keyword used to create a class
# Car → Name of the class (by convention, starts with a capital letter)
# : → Tells Python “the class definition starts here”
## audi = Car()
# audi → Variable name (can be anything)
# = → Assignment operator (store value on the right into the left)
# Car() →
# Car → Class name
# () → Constructor call (creates a new object)
# 👉 Meaning: Create a new object from the Car class and store it in audi.

# Tips 
# Car → Class (Blueprint)
# audi → Object / Instance
# bmw → Object / Instance
# audi and bmw are different objects in memory
# Changing audi will NOT affect bmw

print(type(audi)) # <class '__main__.Car'> # __main__ is the name of the module

print(audi) # <__main__.Car object at 0x000002557D828830> # 0x000002557D828830 is the specific memory address of the object

print(bmw) # <__main__.Car object at 0x000002557D90DA90> # 0x000002557D90DA90 is the specific memory address of the object
```

```python
# Instance Variable and Methods
# Instance Variable is a variable that is associated with an instance of a class. 
# It is a variable that is defined inside a class and can be accessed on an object of that class.
# Instance Method is a method that is defined inside a class and can be called on an object of that class.

audi.tires = 4 
# Instance(Object) created from the Car class ; .(dot operator) used to access or create attributes of an onject)
# tires --> Attribute name(an instance variable) and =(assigmnet operator) 4 --> Value assigned to the attribute
# 👉 Meaning: Create an instance variable called tires inside the audi object and assign the value 4.
audi.windows = 5 
# audi → Same object; . → Attribute access ; windows → Another instance variable = → Assignment operator; 5 → Value assigned
# 👉 Meaning: Create another instance variable called windows for the audi object.

print(audi.windows)
# print → Built-in function to display output;( → Start of function call; audi → Object;
#  . → Access operator; windows → Attribute name; ) → End of function call
# 👉 Meaning: Access the windows value from audi and print it.
print(audi.tires)
```

```python
# Goood Approach to define all the attributes and methods while initializing or while we creating objects itself --> Good Approch
# Instance Variable and Methods

# Creating new class called Dog and defining a constructor
class Dog:
    ## Constructor
    def __init__(self, name, age , color, breed): # def → keyword to define a function; __init__ → constructor method (special method)
# self → reference to the current object; name, age, color → parameters passed while creating an object
# : → starts the function block
        self.name=name
# self.name → variable stored inside the object; = → assignment operator; name → value received from the constructor parameter
# 👉 Meaning: “Store the given name inside this object.”
        self.age=age 
        self.color=color
# Later User can able to add multiple parameters in above construtor 
        self.breed=breed

# Note : The above block of codes is that i am initializing my entire class with a Constructor 
# What happens here?
# ✔ Class Dog is created
# ❌ No object exists (means object yet to create)
# ❌ __init__() is NOT executed



## Creating Objects 
# dog1=Dog() -> dog1 is an object but # TypeError: DOG.__init__() missing 3 required positional arguments: 'name', 'age', and 'color'

# Below creating object and providing all required positional arguments 
dog1 = Dog("buddy",3,"grey","Australian breed") # First it will call the above created Constructor and its 3 parameters (name,age,color)
print(dog1) # creating the .Dog object in specific memory location
# Accessing the object dog1 values
print(dog1.name)
print(dog1.color)

# Lets Create instances of Object in same class Dog
dog2=Dog("Lucy",12,"white","American Breed")
doggy3 = Dog("Bruce",6,"Blacky","China Breed")
# printing the values with calling the object
print(doggy3.name)
print(dog2.breed)
```

```python
# First Creating Class and then Creating Constructor and then Creating Methods and then Objects

# here., parameters such as (name, age, colour, breed ) from constructor are the properties or attributs for the dog

# But the instance methods are functionalities actions or behaviour performed by the Dog ---- eg is below
# eg., dog will Bark ; Dog will catch theif, Dog will Smell good 

# creating class and constructor

class DOG:
    def __init__(self, dog_name, dog_gender):
        self.name=dog_name
        self.gender=dog_gender
    
# Creating instance of methods 
    def bark(self):
        print(f"{self.name} is Barking - woofs") # f-string with self.name (parameters from constructor)

    def catch(self):
        print(f"{self.name} is catched the theif")

# Later User can add multiple methods in future    
    def smell(self):
        print(f"the {self.name} is good at smellings")
    
#Creating the Multiple Objects also user can add objects in future
dog1=DOG("Tommy","Male")
danger_dog2 = DOG("Danger","Female")
# Calling the method name (bark,smell,catch) 
danger_dog2.smell()
dog1.catch()
```
