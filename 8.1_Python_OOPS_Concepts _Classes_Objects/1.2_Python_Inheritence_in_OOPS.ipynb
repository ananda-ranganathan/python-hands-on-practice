```python
# Pyhton - Inheritence in OOPS
# Definition: Inheritence is a mechanism in which one class inherits the properties and methods of another class. 
# This allows you to create a new class that is a specialized version of an existing class. 
# The class that is inherited from is called the parent class, and the class that inherits from it is called the child class. 

# Python supports five types of inheritance: 
# Single: A child class inherits properties and methods from a single parent class. 
# Multiple: A child class inherits properties and methods from multiple parent classes. 
# Multilevel: A child class inherits properties and methods from a parent class, which in turn inherits properties and methods from another parent class.   
# Hierarchical: Multiple child classes inherit properties and methods from a single parent class. 
# Hybrid: A child class inherits properties and methods from both a single parent class and multiple parent classes. 
```

```python
## Single Inheritence in OOPS

## Parent Class

class CAR:
    def __init__(self,windows,engine_type,doors):
        self.windows = windows
        self.engine_type = engine_type
        self.doors=doors
    def drive(self):
        print(f"The person will drive the {self.engine_type} Car")
       # return None

# Note : The print statement doesn’t execute because the class method is never called using an object.

car1 = CAR(4,"petrol",5)
#print(car1.drive())
car1.drive()
```

```python
# Child Class
class Tesla(CAR):
    def __init__(self,windows,engine_type,doors,is_self_driving):
        super().__init__(windows, doors, engine_type) # Consider this as calling the parent class constructor and orders matters 
        self.is_self_driving = is_self_driving
    def self_drive(self):
        return f"Tesla supports self driving: {self.is_self_driving}"
    
car2= Tesla(4,2,"Electric",True) 
# print(car2.self_drive())
car2.self_drive()    # Calling the child class method
car2.drive()

```

```python
# Multiple Inheritence : when a class inherits from more than one base class

# Parent or Base Class 1

class Animal:
    def __init__(self,name):
        self.name=name
    
    def speak(self):
        print(f"Subclass must implement this method")

# Parent or Base Class 2
class Pet:
    def __init__(self,owner):
        self.owner=owner

# Dervied class 
class Dogs(Animal,Pet):
    def __init__(self,name,owner):
        Animal.__init__(self,name)
        Pet.__init__(self,owner)
    
    def speak(self):
        return f"{self.name} says woofs"

# Create an object
dog=Dogs("buddy","Ram")
print(dog.speak())
print(f"Owner Name: {dog.owner} ")
```
