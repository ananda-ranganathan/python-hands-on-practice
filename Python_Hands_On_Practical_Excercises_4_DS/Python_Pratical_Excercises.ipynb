```python
# Simple Calculator

num1 = float(input("Enter the First Number :"))
num2 = float(input("Enter the Second Number :"))
Operations = input("Enter the Operations '+','-','*','/','%'")

# Performing the requested operations 
if Operations == '+':
    print("addtion :", num1 + num2)
elif Operations == '-':
    if num1 >= num2:                              # 10 - 6 = 4
        print("Substaction", num1 - num2)
        if num1<=num2:                                  # 6 - 10 = -4
            print("Negative Subraction", num1 - num2)
    # else:
    #     print("Substraction not working")
elif Operations == '*':
    print("Multiplication", num1 * num2)
elif  Operations == '/':
    if num2 != 0:
        print("Division", num1 / num2)
    else:
        print("Division not posible by Denominator Zero")
elif Operations == '%':
    if num2 != 0:
        print("Modulus", num1 % num2)
    else:
        print("Modulus not posible by Denominator Zero")
else:
    print("enter the input correctly")
```

```python
# While Loop + Continue Keyword
# Example, Printing Odd Numbers from 1 to 10

num = 1
while num <= 10:
    if num % 2 == 0:
        num += 1
        continue
    print(num)
    num += 1

# Code Explaination
# The loop starts with num = 1, which is less than or equal to 10, so the loop condition is true.

# Inside the loop, the if statement checks if num is even (divisible by 2). If it is, num is incremented by 1 using num += 1 
# and the loop continues to the next iteration.

# If num is odd, the print(num) statement is executed, which prints the current value of num. 
# Then, num is incremented by 1 using num += 1.

# This process repeats until num becomes greater than 10, at which point the loop condition is no longer true, and the loop ends.   

```

```python
# Prime Numbers
for num in range(1,55):
    if num>1:
        for i in range(2,num):
            if num%i==0:
                break
        else:
            print(num)
```

```python
# Use a Distionary to count the frequency of elements in List

numbers = [1,2,2,3,3,3,4,4,4,4]
print(numbers)

frequency = {}

for i in numbers:
    if i in frequency:
        frequency[i] = frequency[i] + 1
    else:
        frequency[i] = 1
print(frequency)
```
