```python
# Example 1: Temperature Conversion
def convert_temperature(temp,units):
    # this functions helpful to convert temperature between celsius and Fahrenheit
    if units=='C':
        return (temp * 9/5 + 32 ) # Formula for converting Celsius to Fahrenheit
    elif units=='F':
        return ((temp-32) * 5/9 ) # Formula for converting Fahernheit to Celcius
    else:
        return None 

print(convert_temperature(8,'C'))
print(convert_temperature(77,'F'))
```

```python
# Password Strength Checker 
def is_strong_password(password):
    # this function is check whether the password is strong or not 
    if len(password)<8:
        return False
    if not any (char.isdigit() for char in password):
        return False
    if not any (char.islower() for char in password):
        return False
    if not any (char.isupper() for char in password):
        return False
    if not any (char in ('!@#$%^&*()_+') for char in password):
        return False
    else:
        return True
    
# Calling the function 
print(is_strong_password("WeekPwd"))
print(is_strong_password("An1!23@nDstrngPWd"))
```

```python
# Calculate the Total Cost of an items In a Shopping Cart
def total_cost_in_shpng_cart(cart_items):
    # pass
    total_cost = 0
    for items in cart_items:
        total_cost += items['fruit_price'] * items['fruit_qty']

    return total_cost

# Cart Data

cart_items = [
    { 'fruit_name' : "Apple", 'fruit_price' : 0.4, 'fruit_qty': 3 },
    { 'fruit_name' : "Orange", 'fruit_price' : 0.2, 'fruit_qty': 5 },
    { 'fruit_name' : "Mango", 'fruit_price' : 2.4, 'fruit_qty': 2 },
    { 'fruit_name' : "Jackfruit", 'fruit_price' : 1.2, 'fruit_qty': 5 }
]

# Calling total_cost_in_shpng_cart Fucntion
total_cost = total_cost_in_shpng_cart(cart_items)
print (total_cost)
```

```python
# Checking Palindrome(s)
def Is_palindrome (s):
    s=s.lower().replace(" ","")
    return s==s[::-1]

print (Is_palindrome("A man a Plan a canal PanamA"))
print (Is_palindrome("hellO"))
print (Is_palindrome(""))
```

```python
# Calculate the factorial of a number using recursion
def factorial (n): # n = 6 ; 6*5*4*3*2*1 ; 
    if n==0: 
        return 1
    else:
        return n * factorial(n-1)
    
print(factorial(6))

```

```python
# Counting Frequency of a sentence/words in seperate file 
def count_word_frequence(file_path):
    word_count={}
    with open (file_path, 'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.lower().strip('.,!?";:"\'')
                word_count[word] = word_count.get((word),0) + 1

    return word_count

file_path = 'sample.txt.txt'
word_freq = count_word_frequence(file_path)
print(word_freq)
```
