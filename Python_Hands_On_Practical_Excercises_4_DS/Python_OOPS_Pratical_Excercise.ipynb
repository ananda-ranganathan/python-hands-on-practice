````python
# Modeling a Bank Account

## Define the class for a bank account
class Bank_Account:
    def __init__(self,owner,balance=0):
        self.balance=balance
        self.owner=owner
        
# Creating Instance Methods
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited. New Balance is {self.balance}")
    
    def withdraw(self,amount):
        if amount>self.balance:
            print("Insufficient Funds")
        else:
            self.balance -= amount 
            print(f"{amount} is withdrawn. balance is {self.balance}")
        
    def get_balance(self):
        return self.balance

## Creating an account
account = Bank_Account ("Anand",5000)
print(account.balance)

## Call Instance Methods
account.deposit(100)
account.withdraw(5200)
print(f"Current Balance is {account.get_balance()}")
````

````python
# Most Testcases

class Bank_Account:
    MIN_BALANCE = 1000   # class-level constant

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        # Validation: deposit must be positive
        if amount <= 0:
            print("Deposit amount must be positive")
        else:
            self.balance += amount
            print(f"{amount} deposited. New balance is {self.balance}")

    def withdraw(self, amount):
        # Validation: withdrawal must be positive
        if amount <= 0:
            print("Withdrawal amount must be positive")

        # Validation: minimum balance check
        elif self.balance - amount < Bank_Account.MIN_BALANCE:
            print("Withdrawal denied. Minimum balance of 1000 must be maintained")

        else:
            self.balance -= amount
            print(f"{amount} withdrawn. Balance is {self.balance}")

    def get_balance(self):
        return self.balance
    
# Testcases 
account = Bank_Account("Anand", 5000)

account.deposit(-100)     # ❌ Invalid
account.deposit(500)      # ✅ Valid

account.withdraw(4500)    # ❌ Breaks min balance
account.withdraw(2000)    # ✅ Valid

print(account.get_balance())
````

````python
# Abstraction in Shrink - Payment mode - How Abstraction Takes Place
from abc import ABC, abstractmethod

# 1. Abstract Class
class PaymentGateway(ABC):
    @abstractmethod
    def pay(self, amount):
        pass   # abstraction: only defines WHAT must be done, not HOW

# 2. Concrete Implementation
class UPIPayment(PaymentGateway):
    def validate_upi(self, upi_id):
        # --- Hidden internal detail ---
        return "@" in upi_id   # simple check, user never sees this

    def encrypt_data(self, amount):
        # --- Hidden internal detail ---
        return f"encrypted({amount})"   # fake encryption, hidden from user

    def pay(self, amount):
        # using hidden functions internally
        if not self.validate_upi("ananda@okhdfcbank"):
            return "Payment failed: Invalid UPI ID"
        encrypted = self.encrypt_data(amount)   # hidden encryption step
        # --- Visible output to user ---
        return f"Paid {amount} using UPI"


# =========================
# Demo Usage
# =========================

# Command: create an object of UPIPayment
upi_payment = UPIPayment()

# Command: call the pay method with an amount
result = upi_payment.pay(500)

# Command: print the result (visible to user)
print(result)

# How and Where Abstraction Happens
# At the abstract class (PaymentGateway)
# Defines the contract: every payment must have a pay method.
# Abstraction point: It hides how the payment is processed.
# Inside UPIPayment
# Implements the hidden logic with helper functions like validate_upi and encrypt_data.
# These are internal details — the user never sees them.
# At the pay method
# Calls hidden functions internally.
# Returns only a simple message to the user.

# 👀 What User Sees vs Hidden
# Visible to End User	
# "Paid 500 using UPI"
	
# Hidden Internal Details
# validate_upi() check, encrypt_data() logic
````

````python
# Private Variable in Encapsulation (private Access Modifiers)

# Created the class called Person --> Indentation Starts here 
class Person(): 
    def __init__(self,name,age,gender): 
        self.name=name   
        self.__age=age     
        self.__gender=gender  # Class Person Indentation ends Here 

# the below block of object code is out of Person() class scope
pri_person = Person("Anand",20,"Male") 

print(pri_person.name) 
print(pri_person.gender)  

print(pri_person.age)

# does it correct command so far i am understood the inside and outside within person class?

# Excellent breakdown, Ananda 👌 — you’re on the right track with understanding **inside vs. outside 
# scope** in Python classes. Let’s carefully validate your code and reasoning:


## ✅ Scope & Indentation
# - **Inside the class**: Everything indented under `class Person:` belongs to the class definition. 
# In your code, only the `__init__` method is inside.  
# - **Outside the class**: Once indentation ends, you’re back in global scope. That’s where you create 
# the object `pri_person` and run `print(...)`.

# So yes — your comment annotations about *“Indentation starts here”* and *“Indentation ends here”* are correct.

# ---

## ⚠️ Accessing Attributes
# Now, let’s check the `print` statements:

# ```python
print(pri_person.name)     # ✅ Works → "Anand"
print(pri_person.gender)   # ❌ Error → AttributeError
print(pri_person.age)      # ❌ Error → AttributeError

### Why?
# - `self.name` → stored as `pri_person.name` (public, works fine).  
# - `self.__age` → stored as `pri_person._Person__age` (name‑mangled, not `pri_person.age`).  
# - `self.__gender` → stored as `pri_person._Person__gender` (name‑mangled, not `pri_person.gender`).  

# So your last two print statements fail because those attributes don’t exist under the names you used.


## 🛠 Correct Ways to Access
# Option 1: Use mangled names (not recommended in practice, but works):
# ```python
print(pri_person._Person__age)     # 20
print(pri_person._Person__gender)  # Male
# ```

# Option 2: Use getters (best practice for encapsulation):
# ```python
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.__age = age
        self.__gender = gender

    def get_age(self):
        return self.__age

    def get_gender(self):
        return self.__gender

pri_person = Person("Anand", 20, "Male")

print(pri_person.name)          # Anand
print(pri_person.get_age())     # 20
print(pri_person.get_gender())  # Male
# ```

# ---

## 🔑 Key Understanding
# - Your **scope comments are correct**: object creation and prints are outside the class.  
# - The **commands are not correct** for private variables: `pri_person.age` and `pri_person.gender` 
# don’t exist because of Python’s **name mangling**.  
# - To access them, you need either `_Person__age` / `_Person__gender` or proper getter methods.

# ---

# Would you like me to sketch a **visual diagram of the object layout** 
# (like a box showing `pri_person` with `name`, `_Person__age`, `_Person__gender`) 
# so you can literally see how Python stores these attributes? That would cement the inside/outside scope idea.



# https://copilot.microsoft.com/shares/ePVwTHkn9meW1SBXKdkWA
````
